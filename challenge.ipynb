## Setup — Run This Cell First (not graded)

This cell loads the data and creates the `df` DataFrame you will use throughout the challenge.
**Run it once before starting the exercises.** You should see `Loaded X rows` as output.

```python
import pandas as pd
import plotly.express as px

px.defaults.template = "plotly_white"

# Load both JSON files and combine into one DataFrame
df = pd.concat(
    [
        pd.read_json("StreamingHistory_music_0.json"),
        pd.read_json("StreamingHistory_music_1.json"),
    ],
    ignore_index=True,
)

# Convert endTime from text to a proper datetime object
df["endTime"] = pd.to_datetime(df["endTime"])

# Convert milliseconds to minutes (easier to work with)
df["minutesPlayed"] = df["msPlayed"] / 60_000

print(f"Loaded {len(df):,} rows")
df.head()
```

### Exercise 1 — Summary Statistics (5 pts)

Calculate four headline numbers:
- **`total_hours`** — total listening time in hours
- **`unique_artists`** — number of different artists in the dataset
- **`first_listen`** — the earliest stream date
- **`last_listen`** — the most recent stream date

```python
# Total hours listened  (minutesPlayed column → hours)
total_hours = df["minutesPlayed"].sum() / 60

# Number of unique artists
unique_artists = df["artistName"].nunique()

# Earliest and most recent stream
first_listen = df["endTime"].min()
last_listen = df["endTime"].max()

print(f"Total listening time: {total_hours:,.0f} hours")
print(f"Unique artists:       {unique_artists:,}")
print(f"First listen:         {first_listen:%B %d, %Y}")
print(f"Last listen:          {last_listen:%B %d, %Y}")
```

> **Hint:** `total_hours` should be between 400 and 600.

### Exercise 2 — Top 10 Artists (5 pts)

Find the 10 artists with the most total minutes played.
Store the result in a variable called `top10_artists`.

You'll need to chain four operations: `groupby`, `sum`, `sort_values`, and `head`.

```python
top10_artists = df.groupby("artistName")["minutesPlayed"].sum().sort_values(ascending=False).head(10)

print(top10_artists)
```

### Exercise 3 — Skipped Tracks (5 pts)

A track was **skipped** if it was played for less than 30 seconds (= 30,000 milliseconds).

1. Create a boolean column `df["skipped"]` that is `True` when `msPlayed < 30000`.
2. Calculate `skip_rate` — the fraction of all streams that were skipped (a number between 0 and 1).

```python
# Create the skipped column  (use a comparison operator)
df["skipped"] = df["msPlayed"] < 30_000

# Calculate skip rate (the mean of a boolean column = fraction of True values)
skip_rate = df["skipped"].mean() 

print(f"Skip rate: {skip_rate:.1%}")
```

> **Hint:** `skip_rate` should be between 0.20 and 0.50 (i.e. between 20% and 50%).

### Exercise 4 — Monthly Listening (5 pts)

Resample the data to get the total **minutes played per month**.
Store the result in a variable called `monthly`.

You'll need to set `endTime` as the index first, then use `resample("MS")` (MS = month start).

```python
# Set endTime as the DataFrame index so resample knows what to use
ts = df.set_index("endTime")

# Resample by month start and sum the minutes
monthly = ts["minutesPlayed"].resample("MS").sum()

print(monthly)
```

> **Hint:** `monthly` should have at least 8 rows — one per month covered in the dataset.

### Exercise 5 — Listening by Hour of Day (10 pts)

1. Group the data by **hour of day** (0–23) and sum the minutes played.
   - Use `df["endTime"].dt.hour` to extract the hour from each timestamp.
   - Store the result in a variable called `hourly` with columns `Hour` and `Minutes`.
2. Create a **bar chart** showing minutes listened per hour.
3. In the markdown cell below the code, explain the pattern you see in 2–3 sentences.

```python
#Extract hour of day from timestamp
df["Hour"] = df["endTime"].dt.hour

#Group and add minutes played by hour of day (0-23)
hourly = df.groupby("Hour")["minutesPlayed"].sum()

#Reset and rename index and columns for plotting
hourly = hourly.reset_index()
hourly.columns = ["Hour", "Minutes"]

#Create a new indicator on whether listening is active or inactive based on minutes played (active defined as x > 100 minutes)
hourly["Period"] = hourly["Minutes"].apply(lambda x: "Active" if x > 100 else "Inactive")

#Plot data
fig = px.bar(hourly,
    x="Hour",
    y="Minutes",
    title="Minutes Played by Hour of Day",
    labels={"Hour": "Hour of Day", "Minutes": "Total Minutes Played"},
    color="Period",
    color_discrete_map={"Active": "limegreen", "Inactive": "crimson"})
fig.update_layout(legend_title_text="Key")
fig.show(renderer="browser")
```

**Your explanation:** *(The data shows a clear listening pattern throughout the day, with active listening concentrated between hours 6 and 21, the waking hours when music is most likely intentionally played. Listening peaks around hour 10, dips slightly midday, then rises again in the early afternoon, before tapering off toward evening. The midday dip and afternoon recovery likely reflect breaks in free time, such as meetings, classes, or other activities.)

### Exercise 6 — Most-Skipped Artists (10 pts)

Find the 10 artists with the **highest skip rate**, but only consider artists with at least 15 plays (to avoid noise from artists heard only once).

1. For each artist, calculate total plays and skip rate.
2. Filter to artists with ≥ 15 plays, sort by skip rate descending, and take the top 10.
   - Store the result in a variable called `most_skipped` (10 rows).
3. Create a **horizontal bar chart** showing skip rate per artist.
4. In the markdown cell below, explain in 2–3 sentences what skip rate means and what the results suggest.

*Note: this exercise requires `df["skipped"]` from Exercise 3 — make sure that cell ran first.*

```python
#Aggregrate skip rate and play count per artist, filter to top 10 with at least 15 plays
artist_skips = df.groupby("artistName")["skipped"].agg(plays="count", skip_rate="mean")
filtered = artist_skips[artist_skips["plays"] >= 15].sort_values("skip_rate", ascending=False).head(10).copy()

# Separate Unknown Artist
unknown = filtered[filtered.index == "Unknown Artist"].copy()
known = filtered[filtered.index != "Unknown Artist"].copy()

# Label known artists
known["Most Skipped"] = ["Highest Skip Rate"] + ["Other"] * (len(known) - 1)

# Move Unknown Artist at the end, marked as Excluded
unknown["Most Skipped"] = "Excluded"
most_skipped = pd.concat([known, unknown])

#Plot data
fig = px.bar(most_skipped,
    x="skip_rate",
    y=most_skipped.index,
    orientation="h",
    title="Top 10 Most Skipped Artists (min. 15 plays)",
    labels={"skip_rate": "Skip Rate", "artistName": "Artist"},
    color="Most Skipped",
    color_discrete_map={"Highest Skip Rate": "limegreen", "Other": "cadetblue","Excluded": "lightgray"})
fig.update_layout(xaxis_tickformat=".0%", legend_title_text="Key")
fig.show(renderer="browser")

#Print a table with the same data, formatting skip_rate as a percentage and renaming columns for better readability
most_skipped["skip_rate"] = most_skipped["skip_rate"].map("{:.1%}".format)    
print(most_skipped.rename(columns={"plays": "Plays", "skip_rate": "Skip Rate"}).rename_axis("Artist"))
```

### Exercise 7 — Weekday vs Weekend (10 pts)

Compare average **daily** listening time on weekdays versus weekends.

1. Add a boolean column `df["is_weekend"]` that is `True` for Saturday and Sunday.
   - Use `df["endTime"].dt.dayofweek` — Monday = 0, Sunday = 6.
2. Calculate the average daily listening minutes separately for weekdays and weekends.
   - Store results in variables `avg_weekday` and `avg_weekend`.
3. Create a **visualization** that clearly compares the two.
4. In the markdown cell below, explain the difference in 2–3 sentences.

```python
#Differentiate between weekday and weekend (0-6), where 5 and 6 are Saturday and Sunday
df["is_weekend"] = df["endTime"].dt.dayofweek >= 5

#Group by date, then calculate minutes, and determine type of day it is
daily = df.groupby(df["endTime"].dt.date).agg(
    minutes=("minutesPlayed", "sum"),
    is_weekend=("is_weekend", "first"))

#Filter daily minutes, and calculate averages
avg_weekday = daily[~daily["is_weekend"]]["minutes"].mean()
avg_weekend = daily[daily["is_weekend"]]["minutes"].mean()

# Plot the data
fig = px.bar(
    x=["Weekday", "Weekend"],
    y=[avg_weekday, avg_weekend],
    title="Average Daily Listening: Weekday vs Weekend",
    labels={"x": "Day Type", "y": "Avg Minutes per Day"},
    color=["Weekday", "Weekend"],
    color_discrete_map={"Weekday": "goldenrod", "Weekend": "rebeccapurple"})
fig.update_layout(showlegend=False)
fig.show(renderer="browser")

#Print a table with the data, rounding average minutes to 1 decimal place and renaming columns for better readability
summary = pd.DataFrame({
    "Day Type": ["Weekday", "Weekend"],
    "Avg Minutes": [round(x, 1) for x in [avg_weekday, avg_weekend]]
})
print(summary)
```
